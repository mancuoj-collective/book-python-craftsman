```python
def extract_values(s):
    items = s.split(":")
    if len(items) == 2:
        return items[1]
    else:
        return ""


print(extract_values("name:test"))
print(extract_values("1"))


def extract_value_v2(s):
    return s.partition(":")[-1]


print(extract_value_v2("name:test"))
print(extract_value_v2("1"))
```
