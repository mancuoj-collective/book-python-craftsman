```python
def extract_values(s):
    items = s.split(":")
    if len(items) == 2:
        return items[1]
    else:
        return ""


print(extract_values("name:test"))
print(extract_values("1"))


def extract_value_v2(s):
    return s.partition(":")[-1]


print(extract_value_v2("name:test"))
print(extract_value_v2("1"))
```

```python
# 增强代码可读性
from enum import Enum

DAILY_POINTS_REWARD = 100
VIP_EXTRA_REWARD = 20


class UserType(Enum):
    VIP = 3
    BANNED = 13


def add_daily_points(user):
    if user.type == UserType.BANNED:
        return
    if user.type == UserType.VIP:
        user.points += VIP_EXTRA_REWARD
    user.points += DAILY_POINTS_REWARD
```
